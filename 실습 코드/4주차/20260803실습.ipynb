```python
import requests

items_id = int(input("id 입력"))
url = f'http://127.0.0.1:8000/item/{items_id - 1}'

res = requests.get(url)

if res.status_code == 200:
  data = res.json()
  print(f"Response : {res.status_code}\nInput Data : {items_id}")
  print(data)
  print(f"item_name : {data['item']['item_name']} // price : {data['item']['price']}")
else:
  print(f"Error : {res.status_code} - {res.text}")
  data = ""
```
