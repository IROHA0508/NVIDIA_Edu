- mysql 연결

```python
import pymysql

connection = pymysql.connect(
    host='127.0.0.1',   # localhost, 서버 주소
    user='root',        # 사용자
    password='0000',  # 패스워드
    database='mysqldb', # 사용할 DB
    charset='utf8mb4'
)
```

- CRUD 구현

```python
# 데이터 추가(insert)

with connection.cursor() as cursor:
    # 쿼리 작성
    sql = """
    insert into menu (name, price, description)
    values (%s, %s, %s)
    """

    # sql 실행
    cursor.execute(
        sql, 
        ("불고기 피자", 19000, "한국식 불고기 피자입니다.")
    )

# DB에 실행 결과 반영
connection.commit()
print("메뉴가 추가되었습니다.") 
```

```python
# 데이터 조회(Select)

import pandas as pd

with connection.cursor() as cursor:
    # sql 생성
    sql = 'select * from menu'
    cursor.execute(sql)

    # 조회된 전체 데이터 가져오기
    menus = cursor.fetchall()
    print('==== 메뉴 출력 ====')

    for menu in menus:
        print(menu)

    df = pd.DataFrame(menus, columns=['id', 'name', 'price', 'ddscription'])

df
```

```python
with connection.cursor() as cursor:
    # sql 생성
    sql = 'select name, price from menu where id=%s'
    cursor.execute(sql, 2)

    # 조회된 전체 데이터 가져오기
    menus = cursor.fetchall()
    print('==== 메뉴 출력 ====')

    for menu in menus:
        print(menu)
```

```python
# 데이터 수정(Dudate)
with connection.cursor() as cursor:
    sql = """
        update menu
        set price = %s,
            description=%s
        where id = %s
    """

    # id=2 => 메뉴 가격을 50000 변경
    cursor.execute(sql, (50000, '프러플 향을 첨가한 크림 파스타입니다.', 2))

    #  DB 반영
    connection.commit()
```

```python
# 데이터 삭제(Delete)
with connection.cursor() as cursor:
    sql = """
    delete from menu 
    where id=%s
    """
    cursor.execute(sql, (4,))

connection.commit()

print('삭제 완료')
```
